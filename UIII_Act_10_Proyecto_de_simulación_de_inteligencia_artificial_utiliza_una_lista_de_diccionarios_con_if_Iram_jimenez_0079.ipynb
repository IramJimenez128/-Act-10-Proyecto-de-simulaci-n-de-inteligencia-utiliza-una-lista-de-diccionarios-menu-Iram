```python
# --- SISTEMA DE IA: ASISTENTE DE REDACCIÓN V3.0 ---

# Inicialización de las estructuras solicitadas
# 'periodista' es la lista que almacenará los diccionarios [2, 8]
periodista = []
# 'los_periodistas' es el diccionario temporal que inicia vacío [3]
los_periodistas = {}

print("--- IA NEWS-BOT: GESTIÓN DE PLANTILLA ---")

# Ciclo infinito para el menú interactivo [4]
while True:
    print("\n--- MENÚ PRINCIPAL ---")
    print("1. Registrar nuevo periodista")
    print("2. Listar contenido de 'periodista'")
    print("3. IA: Análisis de misiones")
    print("4. Salir")

    # Captura de la opción del usuario [9, 10]
    opcion = input("\nSeleccione una opción (1-4): ")

    # Rama 1: Agregar un diccionario a la lista [5]
    if opcion == "1":
        nombre = input("IA: Ingrese nombre: ")
        especialidad = input("IA: Ingrese especialidad (politica/deportes/cultura): ")

        # Se rellena el diccionario 'los_periodistas' con los datos [3]
        los_periodistas = {
            "nombre": nombre,
            "especialidad": especialidad.lower()
        }

        # Se añade el diccionario a la lista global 'periodista' [8]
        periodista.append(los_periodistas)
        print(f"IA: {nombre} ha sido guardado en la lista.")

    # Rama 2: Mostrar el listado usando un ciclo for [6]
    elif opcion == "2":
        if not periodista: # Verifica si la lista está vacía [8]
            print("IA: La lista 'periodista' no tiene registros aún.")
        else:
            print("\n--- REGISTROS EN LA LISTA ---")
            # El ciclo for recorre cada diccionario dentro de la lista [7]
            for p in periodista:
                print(f"Ficha -> Nombre: {p['nombre']} | Área: {p['especialidad']}")

    # Rama 3: Lógica de IA para el análisis de los datos [11]
    elif opcion == "3":
        if not periodista:
            print("IA: No hay datos suficientes para el análisis.")
        else:
            print("IA: Ejecutando escaneo de perfiles...")
            for p in periodista:
                print(f"\n>> Recomendación para {p['nombre']}:")

                # Evaluación lógica basada en el contenido del diccionario [12]
                if p['especialidad'] == "politica":
                    print("IA: Prioridad CRÍTICA. Enviar a cubrir el Congreso.")
                elif p['especialidad'] == "deportes":
                    print("IA: Prioridad ALTA. Solicitar entrevista en el estadio.")
                elif p['especialidad'] == "cultura":
                    print("IA: Prioridad MEDIA. Redactar nota sobre el museo.")
                else:
                    print("IA: Categoría genérica. Asignar a mesa de noticias.")

    # Rama 4: Finalización del sistema [13]
    elif opcion == "4":
        print("IA: Cerrando base de datos. ¡Hasta pronto!")
        break # Rompe el ciclo while True

    # Manejo de errores del menú [14]
    else:
        print("IA: Error. La opción ingresada no es válida.")

print("\n--- PROGRAMA FINALIZADO ---")
print("Porgrama realizado por Iram Jimenez 2H 0079")
```
