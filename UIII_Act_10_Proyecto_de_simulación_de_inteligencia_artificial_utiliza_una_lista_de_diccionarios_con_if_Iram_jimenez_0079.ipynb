```python
# --- SISTEMA DE IA: ASISTENTE DE REDACCIÓN PERIODÍSTICA V2.0 ---

# Inicialización de estructuras de datos vacías
lista_periodistas = []  # Lista que guardará los diccionarios [1]
periodista_actual = {}  # Diccionario temporal vacío [2]

print("--- IA NEWS-BOT: INICIANDO ASISTENTE VIRTUAL ---")
print("IA: Hola. Soy tu asistente inteligente de redacción.")

# Ciclo infinito para el menú interactivo
while True:
    print("\n--- MENÚ PRINCIPAL DEL EDITOR ---")
    print("1. Registrar nuevo periodista")
    print("2. Mostrar plantilla del equipo (Listar)")
    print("3. IA: Asignar misiones especiales (Análisis)")
    print("4. Salir")

    opcion = input("\nEditor, seleccione una opción (1-4): ")

    # Rama 1: Agregar un periodista a la lista de diccionarios
    if opcion == "1":
        # Se capturan datos para crear el perfil
        nombre = input("IA: Ingrese el nombre del periodista: ")
        especialidad = input("IA: Ingrese especialidad (politica/deportes/cultura): ")

        # Se rellena el diccionario con la información [3]
        periodista_actual = {
            "nombre": nombre,
            "especialidad": especialidad.lower()
        }

        # Se agrega el diccionario a la lista global [1]
        lista_periodistas.append(periodista_actual)
        print(f"IA: {nombre} ha sido agregado exitosamente a la plantilla.")

    # Rama 2: Listar periodistas utilizando un ciclo for
    elif opcion == "2":
        if not lista_periodistas: # Verifica si la lista está vacía [4]
            print("IA: La lista de periodistas está vacía actualmente.")
        else:
            print("\n--- PLANTILLA DE PERIODISTAS ---")
            # Ciclo for para recorrer la lista de diccionarios [5]
            for p in lista_periodistas:
                print(f"Periodista: {p['nombre']} | Área: {p['especialidad']}")

    # Rama 3: Simulación de IA para asignar misiones (Lógica condicional)
    elif opcion == "3":
        if not lista_periodistas:
            print("IA: No hay periodistas registrados para analizar.")
        else:
            print("IA: Analizando perfiles para asignar misiones...")
            # El ciclo for permite que la IA procese a cada periodista de la lista
            for p in lista_periodistas:
                print(f"\n>> Misión para {p['nombre']}:")

                # Estructuras elif para determinar el consejo de la IA [6]
                if p['especialidad'] == "politica":
                    print("IA CONSEJO: Dirígete al Palacio de Gobierno. Se rumorea un cambio de gabinete.")
                elif p['especialidad'] == "deportes":
                    print("IA CONSEJO: Ve al estadio central. El entrenamiento de hoy es a puerta cerrada.")
                elif p['especialidad'] == "cultura":
                    print("IA CONSEJO: Cubre la inauguración de la galería nacional. Busca al artista principal.")
                else:
                    print("IA CONSEJO: Especialidad genérica detectada. Monitorea las redes sociales por noticias de última hora.")

    # Rama 4: Finalizar el programa
    elif opcion == "4":
        print("IA: Guardando cambios... Sesión finalizada. ¡Buen cierre de edición!")
        break # Rompe el ciclo while True para salir [7]

    # Manejo de opciones no válidas
    else:
        print("IA: Error. Esa opción no existe en mi base de datos.")

print("\n--- SISTEMA APAGADO ---")
print("Programa realizado por Iram Jimenez 0079")
```
